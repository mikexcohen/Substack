# Figure 1: Examples of transformations

```python
# the vector
v = np.array([1,2])[:,None]

# transformation matrices
T1 = np.array([[1,0],[0,2]])
T2 = np.array([[1,1],[0,1]])
T3 = np.array([[-1,0],[.5,1]])


# create a figure
fig,axs = plt.subplots(1,3,figsize=(9,3))

# loop over matrices
for ax,T in zip(axs.flatten(),[T1,T2,T3]):

    # apply the transformation
    y = T@v

    # and visualize
    ax.plot([0,v[0][0]],[0,v[1][0]],'w',linewidth=2,label='Original')
    ax.plot([0,y[0][0]],[0,y[1][0]],'m--',label='Transformed')
    ax.set(xlim=[-4,4],ylim=[-4,4])
    ax.grid(linestyle='--',color='w',linewidth=.3)
    ax.legend(fontsize=8)

    # add the matrix
    Ttxt = f'T = [[{T[0,0]}, {T[0,1]}],\n     [{T[1,0]}, {T[1,1]}]]'
    ax.text(.05,.05,Ttxt,transform=ax.transAxes,verticalalignment='bottom',family='monospace',
            fontsize=11,bbox=dict(facecolor='k') )


plt.tight_layout()
plt.show()
```

# Figure 2: Examples of pure rotations

```python
# the vector
v = np.array([1,2])[:,None]

# transformation angles (matrices created in the loop)
angles = [ np.pi/3, 5*np.pi/6, 21*np.pi/7 ]


# create a figure
fig,axs = plt.subplots(1,3,figsize=(9,3))

# loop over matrices
for ax,angle in zip(axs.flatten(),angles):

    # create the Givens matrix
    T = np.array([ [ np.cos(angle), np.sin(angle)],
                   [-np.sin(angle), np.cos(angle)]] )

    # apply the transformation
    y = T@v

    # and visualize
    ax.plot([0,v[0][0]],[0,v[1][0]],'w',linewidth=2,label='Original')
    ax.plot([0,y[0][0]],[0,y[1][0]],'m--',label='Transformed')
    ax.set(xlim=[-4,4],ylim=[-4,4])
    ax.grid(linestyle='--',color='w',linewidth=.3)
    ax.legend(fontsize=8)

    # add the matrix
    Ttxt = f'$\\theta$ = {angle:.2f} rad.'
    ax.text(.05,.05,Ttxt,transform=ax.transAxes,verticalalignment='bottom',family='monospace',
            fontsize=11,bbox=dict(facecolor='k') )


plt.tight_layout()
plt.show()
```

# Find the Givens angle

```python
# find the Givens angle
M = np.array([ [1,2],[2,5] ])

theta = .5 * np.atan2(2*M[0,1],M[1,1]-M[0,0])
print(f'The Givens angle is {theta:.3f} rad.')
```

```python
# create the Givens matrix
G = np.array([ [np.cos(theta),np.sin(theta)],
               [-np.sin(theta),np.cos(theta)] ])

# apply to the matrix
Mp = G.T @ M @ G

# let's have a look!
Mp
```

```python
# visualize
fig,axs = plt.subplots(1,3,figsize=(8,3))

h = axs[0].imshow(M,vmin=0,vmax=6)
axs[0].set(title='$\\mathbf{M}$',xticks=[],yticks=[])
fig.colorbar(h,ax=axs[0],pad=.02,fraction=.047)

h = axs[1].imshow(G,vmin=-.2,vmax=.6,cmap='plasma')
axs[1].set(title='$\\mathbf{G}$',xticks=[],yticks=[])
fig.colorbar(h,ax=axs[1],pad=.02,fraction=.047)

h = axs[2].imshow(Mp,vmin=0,vmax=5)
axs[2].set(title="$\\mathbf{M'}$",xticks=[],yticks=[])
fig.colorbar(h,ax=axs[2],pad=.02,fraction=.047)

plt.tight_layout()
plt.show()
```

```python
# Givens rotation on a 2x2 symmetric matrix
def givens(M):
  theta = np.atan2(2*M[0,1],M[1,1]-M[0,0]) / 2
  c = np.cos(theta)
  s = np.sin(theta)
  return np.array([ [c,s],[-s,c] ])


# one Jacobi sweep
def jacobi_sweep(Mprime):

  # matrix size
  m = Mprime.shape[0]

  # loop over upper-triangle indices
  for i,j in zip(*np.triu_indices(m,1)):

    # get givens matrix for submatrix
    submat = Mprime[np.ix_([i,j],[i,j])]
    Gsub = givens(submat)

    # create full Givens matrix
    G = np.eye(m)
    G[np.ix_([i,j],[i,j])] = Gsub

    # apply to the matrix
    Mprime = G.T @ Mprime @ G

  # return the final rotated matrix
  return Mprime
```

```python
# an example 4x4 matrix
matrix = np.array([ [1,2,3,4],
                    [2,5,6,7],
                    [3,6,8,9],
                    [4,7,9,10] ]).astype(np.float64)

# one jacobi rotation
Mprime = jacobi_sweep(matrix.copy())
print(np.array2string(Mprime, precision=2, suppress_small=True))
```

```python
# run multiple Jacobi sweeps
niter = 10

matrix_norms = np.zeros(niter)
offdiag_norms = np.zeros(niter)

# initial copy of the matrix
Mprime = matrix.copy()

# loop over iterations
for iter in range(niter):

  # run a sweep
  Mprime = jacobi_sweep(Mprime)

  # calculate some metrics
  matrix_norms[iter] = np.linalg.norm(Mprime,'fro')
  only_offdiag = Mprime - np.diag(np.diag(Mprime))
  offdiag_norms[iter] = np.linalg.norm(only_offdiag,'fro')


_,axs = plt.subplots(1,2,figsize=(10,3))
axs[0].plot(matrix_norms,'ks',markerfacecolor=[.7,.9,.7],markersize=10)
axs[0].set(xlabel='Sweeps',ylabel='Frobenius norm',title="Norm of rotated matrix")

axs[1].plot(offdiag_norms,'ko',markerfacecolor=[.9,.7,.7],markersize=10)
axs[1].set(xlabel='Sweeps',ylabel='Frobenius norm',yscale='log',title='Norm of off-diagonals')

print(np.array2string(Mprime, precision=3, suppress_small=True))
```
