# **Figure 1: Examples of transformations**

```python
# the vector
v = np.array([1,2])[:,None]

# transformation matrices
T1 = np.array([[1,0],[0,2]])
T2 = np.array([[1,1],[0,1]])
T3 = np.array([[-.1,0],[.2,.2]])


# create a figure
fig,axs = plt.subplots(1,3,figsize=(9,3))

# loop over matrices
for ax,T in zip(axs.flatten(),[T1,T2,T3]):

    # apply the transformation
    y = T@v

    # and visualize
    ax.plot([0,v[0][0]],[0,v[1][0]],'w',linewidth=2,label='Original')
    ax.plot([0,y[0][0]],[0,y[1][0]],'m--',label='Transformed')
    ax.set(xlim=[-4,4],ylim=[-4,4])
    ax.grid(linestyle='--',color='w',linewidth=.3)
    ax.legend(fontsize=8)

    # add the matrix
    Ttxt = f'T = [[{T[0,0]}, {T[0,1]}],\n     [{T[1,0]}, {T[1,1]}]]'
    ax.text(.05,.05,Ttxt,transform=ax.transAxes,verticalalignment='bottom',family='monospace',
            fontsize=11,bbox=dict(facecolor='k') )


plt.tight_layout()
plt.show()
```

# **Figure 2: Example of Householder transformations**

```python
# normalized vector to flip across
u = np.array([ [1],[2] ],dtype=float)
u /= np.linalg.norm(u)

# to-flip vector
w = np.array([ [-1],[3] ],dtype=float)

# householder matrix
H = np.eye(2) - 2*u@u.T

# rotated vector
Hw = H@w


# draw the vector and its perpendicular
plt.plot([0,u[0].item()],[0,u[1].item()],linewidth=2,label='$\\mathbf{u}$')
plt.axline((0,0),slope=-(u[0].item()/u[1].item()),
           color='w',linestyle='--',linewidth=.7,
           label='$\\mathbf{u}_{\\perp}$')

# the vector and its rotation
plt.plot([0,w[0].item()],[0,w[1].item()],'r',linewidth=2,label='$\\mathbf{w}$')
plt.plot([0,Hw[0].item()],[0,Hw[1].item()],'r--',linewidth=2,label='$\\mathbf{Hw}$')

# finalize the plot
plt.gca().set_box_aspect(1)
lims = max(abs(w).max(),abs(Hw).max(),abs(u).max()) + .1
plt.gca().set(xlim=[-lims,lims],ylim=[-lims,lims],xlabel='x',ylabel='y')
plt.legend()
plt.grid(linewidth=.3)

plt.show()
```

# **Example in a 2x2 matrix**

```python
# the matrix
A = np.array([ [1,3],[2,1] ],dtype=float)

# Step 1: the column and row (pivot) index
i = 0

# Step 2: define beta
a = A[:,[i]] # extract the column for convenience
beta = np.linalg.norm(a) * -np.sign(a[i])

# Step 3: create u
v = a.copy()
v[0,0] -= beta.item()
u = v / np.linalg.norm(v)

# Step 4: create the Householder matrix
H = np.eye(2) - 2*u@u.T

# Step 5: Create the new A1 matrix
A1 = H@A
```

# **Sweep through the columns for larger matrix**

```python
# a function to handle the single-column reflection
def householder(A,i):

  # Step 2: define beta
  a = A[i:,[i]] # extract the column starting from the pivot row
  beta = np.linalg.norm(a) * -np.sign(a[0])

  # Step 3: create u
  v = a.copy()
  v[0,0] -= beta.item()
  u = v / np.linalg.norm(v)

  # Step 4: create the Householder matrix
  H = np.eye(len(v)) - 2*u@u.T

  # Step 5: Create the new A1 matrix
  A[i:,:] = H@A[i:,:]

  # resize the H
  Hfull = np.eye(A.shape[0])
  Hfull[i:,i:] = H

  return A,Hfull
```

```python
# now to implement QR decomposition using Householder reflection

# a matrix and a copy for R
M = np.random.randint(-10,11,size=(5,4)).astype(float)
R = M.copy()

# initialize Q storage
Qs = []

# loop over the rows/cols
for i in range(min(M.shape)):
  R,H = householder(R,i)
  Qs.append(H)

# create Q as products of H's
Q = np.linalg.multi_dot(Qs)
```

```python
# confirm against NumPy's QR function
Q_np,R_np = np.linalg.qr(M,mode='complete')

print('My R:\n',np.array2string(R, precision=2, suppress_small=True))
print('\nNumPy R:\n',np.array2string(R_np, precision=2, suppress_small=True))

print('\nMy Q:\n',np.array2string(Q, precision=2, suppress_small=True))
print('\nNumPy Q:\n',np.array2string(Q_np, precision=2, suppress_small=True))
```

```python
# confirm A = QR
np.allclose(M,Q@R)
```

```python
# confirm Q'Q = I
np.allclose(Q.T@Q,np.eye(Q.shape[0]))
```

```python
_,axs = plt.subplots(1,len(Qs),figsize=(10,3))

for i in range(len(Qs)):
  axs[i].imshow(Qs[i],vmin=-1,vmax=1,cmap='plasma')
  axs[i].set(title=f'H from iteration {i}')

plt.tight_layout()
plt.show()
```
